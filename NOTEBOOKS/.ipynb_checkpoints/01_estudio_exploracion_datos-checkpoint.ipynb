## 8. Filtrar datos

¿Qué estudiantes obtuvieron 90 o más en matemáticas?

```python
df[df["math score"] >= 90]
```

```python
(df["math score"] >= 90).sum()
```

## 9. Agrupar datos

```python
df.groupby("test preparation course")[["math score", "reading score", "writing score"]].mean()
```

### Importante

Una diferencia observada entre grupos no demuestra causalidad. Por ahora estamos realizando **análisis exploratorio**.

## 10. Primera visualización

```python
df["math score"].plot(kind="hist", bins=10)
plt.title("Distribución de puntuaciones de Matemáticas")
plt.xlabel("Puntuación")
plt.ylabel("Número de estudiantes")
plt.show()
```

## 11. Comparar promedios por materia

```python
promedios = df[["math score", "reading score", "writing score"]].mean()
promedios.plot(kind="bar")
plt.title("Promedio por materia")
plt.ylabel("Puntuación")
plt.xlabel("Materia")
plt.show()
```
